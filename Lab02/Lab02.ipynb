## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# (a) seed 2026, one random number in [0, 1)
np.random.seed(2026)
print("T3a: one rand value =", np.random.rand())

# (b) 25 numbers from the standard normal distribution
normal25 = np.random.randn(25)
print("T3b: 25 randn values =\n", np.round(normal25, 4))

# (c) seed 42, roll 10,000 Ludo dice, count each face
np.random.seed(42)
rolls = np.random.randint(1, 7, size=10000)
faces, counts = np.unique(rolls, return_counts=True)
print("T3c: faces =", faces)
print("T3c: counts =", counts)

# (d) estimate P(six) and compare with 1/6
p_six = (rolls == 6).mean()
print("T3d: P(six) estimated =", round(p_six, 4), "| theoretical 1/6 =", round(1/6, 4),
      "| difference =", round(abs(p_six - 1/6), 4))

# (e) consecutive double sixes (no loops): compare each roll with the next one
is_six = rolls == 6
double_sixes = np.sum(is_six[:-1] & is_six[1:])
print("T3e: times two sixes came in a row =", double_sixes)
```

Using `units`, `pop_2023` and `pop_2017`:

## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# (a) shape of tmax
print("T7a: tmax shape =", tmax.shape)

# (b) annual mean maximum temperature per city (axis=1 collapses the months)
annual_mean = tmax.mean(axis=1)
print("T7b: annual mean max temp (°C):")
for c, v in zip(cities, annual_mean):
    print(f"      {c:<10s} {v:5.2f} °C")

# (c) hottest month of each city
hot_idx = tmax.argmax(axis=1)
print("T7c: hottest month per city =", dict(zip(cities.tolist(), months[hot_idx].tolist())))
print("T7c: peak temperature (°C)   =", dict(zip(cities.tolist(), tmax.max(axis=1).tolist())))

# (d) 5-city average for each month (axis=0 collapses the cities)
print("T7d: 5-city average max temp per month (°C):")
print(dict(zip(months.tolist(), np.round(tmax.mean(axis=0), 2).tolist())))

# (e) number of months above 35 °C in each city (mask + sum over months)
n_hot = (tmax > 35).sum(axis=1)
print("T7e: months above 35 °C per city =", dict(zip(cities.tolist(), n_hot.tolist())))

# (f) Lahore in °F
lahore_f = tmax[0] * 9/5 + 32
print("T7f: Lahore max temp (°F) =", np.round(lahore_f, 1))

# (g) monsoon rain (Jul-Sep = columns 6:9) per city
monsoon = rain[:, 6:9].sum(axis=1)
print("T7g: monsoon (Jul-Sep) rain per city (mm) =", dict(zip(cities.tolist(), np.round(monsoon, 1).tolist())))

# (h) wettest monsoon city and monsoon share of the yearly rain
print("T7h: wettest monsoon city =", cities[monsoon.argmax()], "with", round(monsoon.max(), 1), "mm")
yearly = rain.sum(axis=1)
share_monsoon = monsoon / yearly * 100
print("T7h: yearly rain per city (mm) =", dict(zip(cities.tolist(), np.round(yearly, 1).tolist())))
print("T7h: % of yearly rain that falls in the monsoon:")
for c, s in zip(cities, share_monsoon):
    print(f"      {c:<10s} {s:5.1f} %")
```

**Interpretation (T7):** The biggest surprise was Quetta. It is the only city whose hottest month is July (36.8 °C), and only 17.9 mm of its rain (6 %) falls in the monsoon, because its rain comes in winter and spring instead. Islamabad receives 763 mm in Jul–Sep, far more than Lahore (454 mm), even though the two cities are only a short distance apart. Also, Karachi has the highest annual mean maximum (32.5 °C) although its peak (35.7 °C) is lower than Peshawar's (39.6 °C), because its winters are warm and its temperature stays steady all year.

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)

- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# (a) average inflation of each year (axis=1 = across the 10 months)
avg = cpi.mean(axis=1)
print("T8a: average inflation FY2024 =", round(avg[0], 2), "% | FY2025 =", round(avg[1], 2), "%")

# (b) FY2024 months with inflation above 25 %
print("T8b: FY2024 months with inflation > 25 % =", cpi_months[cpi[0] > 25])

# (c) drop in inflation per month (FY2024 - FY2025), in percentage points
drop = cpi[0] - cpi[1]
print("T8c: drop per month (percentage points) =", np.round(drop, 1))

# (d) month with the biggest drop
print("T8d: biggest drop in", cpi_months[drop.argmax()], "=", round(drop.max(), 1), "percentage points")

# (e) label FY2025 months: Low (< 5), Moderate (5-10), High (>= 10)
label = np.where(cpi[1] < 5, "Low", np.where(cpi[1] < 10, "Moderate", "High"))
print("T8e: FY2025 labels =", dict(zip(cpi_months.tolist(), label.tolist())))
```

Prices were still rising in every month, since inflation stayed above zero, but the pace slowed from very high (above 25 % in seven FY2024 months) to almost flat (0.3 % in April 2025). Prices did not fall; they simply stopped rising so quickly.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# (a) add 5 grace marks, capped at 100
after = np.clip(marks + 5, 0, 100)
print("T9a: marks before =", marks)
print("T9a: marks after  =", after)

# (b) pass % (>= 50) after grace
pass_pct = (after >= 50).mean() * 100
print("T9b: pass percentage after grace =", pass_pct, "%")

# (c) students who passed ONLY because of grace (failed before, pass after)
only_grace = np.where((marks < 50) & (after >= 50))[0]
print("T9c: indices of students saved by grace =", only_grace, "| their marks:", marks[only_grace], "->", after[only_grace])

# (d) z-score of each student after grace
z = (after - after.mean()) / after.std()
print("T9d: z-scores =", np.round(z, 3))
print("T9d: check mean =", round(z.mean(), 6), "| std =", round(z.std(), 6))
```

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
import matplotlib.pyplot as plt

# (a) draw the "7" (white = 255, grey = 128)
img = np.zeros((8, 8), dtype=int)
img[1, 1:7] = 255       # top bar
img[2:7, 5] = 255       # diagonal-ish stem
img[4, 3:7] = 128       # grey crossbar
print("T10a: original image:\n", img)

# (b) flip left-right
flipped = img[:, ::-1]
print("T10b: flipped left-right:\n", flipped)

# (c) invert
inverted = 255 - img
print("T10c: inverted:\n", inverted)

# (d) crop rows 1-4 and columns 3-6 (stop excluded)
cropped = img[1:5, 3:7]
print("T10d: cropped (rows 1-4, cols 3-6):\n", cropped)

# (e) brighten by 80, capped at 255
bright = np.clip(img + 80, 0, 255)
print("T10e: brightened +80 (capped at 255):\n", bright)

# (f) shape, dtype, number of white pixels
print("T10f: shape =", img.shape, "| dtype =", img.dtype, "| white pixels =", (img == 255).sum())

# Optional: display all versions
fig, ax = plt.subplots(1, 5, figsize=(13, 3))
for a, im, t in zip(ax, [img, flipped, inverted, cropped, bright],
                    ["Original", "Flipped", "Inverted", "Cropped", "Brightened"]):
    a.imshow(im, cmap="gray", vmin=0, vmax=255); a.set_title(t); a.axis("off")
plt.show()
```
