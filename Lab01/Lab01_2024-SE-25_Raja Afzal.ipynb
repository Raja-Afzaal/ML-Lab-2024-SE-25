Q6) What is the main difference between a tuple and a list?

Answer:
1. Lists use square brackets [] and are MUTABLE (their elements can be modified, added, or removed after creation).
2. Tuples use parentheses () and are IMMUTABLE (their elements CANNOT be modified or changed once created).

Q7) Function: grab the domain from "user@domain.com".

```python
def domainGet(email):
    return email.split('@')[1]
print("Output:", domainGet('user@domain.com')) 
```

Q8) Function: return True if the word "dog" is in a string (case-insensitive).

```python
def findDog(s):
    return 'dog' in s.lower()

print("Output:", findDog('Is there a Dog here?'))
```

Q9) Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def countDog(s):
    return s.lower().count('dog')

print("Output:", countDog('This dog runs faster than the other dog dude!')) 
```

Q10) caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5     

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
print("Output:")
print(caught_speeding(81, True))    
print(caught_speeding(81, False))   
print(caught_speeding(60, False))   
```
